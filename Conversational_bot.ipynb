```python
# Client və dialoq tarixçəsinin yaradılması
messages = [
    {"role": "system", "content": "You are a friendly conversational assistant."}
]

# # introduce yourself
messages.append({"role": "user", "content": "Hello! My name is Aybeniz."})

response1 = openrouter_client.chat.completions.create(
    model="openrouter/auto",
    messages=messages
)

bot_reply1 = response1.choices[0].message.content
messages.append({"role": "assistant", "content": bot_reply1})
print("Bot:", bot_reply1)

# # check if it remembers what you have said (for example your name)
messages.append({"role": "user", "content": "What is my name?"})

response2 = openrouter_client.chat.completions.create(
    model="openrouter/auto",
    messages=messages
)

bot_reply2 = response2.choices[0].message.content
print("\nBot:", bot_reply2)
```

```python
messages = []
print("Write a prompt to initiate chat.")
while True:
    user_input = input("If you want to end chat write 'quit'.")
    messages.append({"role":"user", "content":user_input})
    
    if user_input == "quit":
        break
    response = openrouter_client.responses.create(model='openrouter/auto',
                                                  input = messages,
                                                  instructions="Always ask if user wants to ask something else. Keep your answer short.",
                                                  tools=[
                                                      { "type": "web_search" }
                                                  ]
                                                 )
    
    messages.append({"role":"assistant", "content":response.output_text})
    print(response.output_text)
```
